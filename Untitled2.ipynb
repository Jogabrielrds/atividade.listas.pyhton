```python
#vamos jogar um jogo gen
#temos que contar de um 1 a 100
#se o numero é divisivel por 3, escreve FIZZ
#se for divisivel por 5, escreve BUZZ
#se for divisivel por ambos, escreve FIZZBUZZ

for numero in range(1,20):

 if numero % 3== 0 and numero % 5 == 0:
  print("FIZZBUZZ")

 elif numero % 3 ==0:
   print("FIZZ")

 elif numero % 5 ==0:
    print("BUZZ")

 else:
    print(numero)
```

```python
#entendendo o comando for

lista = ["a "," b"," c"," d"]
len(lista) #len = comando para descobrir a quantidade de elementos de uma lista


#esse foi um for unitario

lista1 = ["a","b","c","d"]
lista2 = [1,2,3,4]

for i,j in zip(lista1,lista2):
  print(f" o valor da lista1 é {i}")
  print(f" o valor da lista2 é {j}")

len(lista1)
```

```python
produtos= ["Camisa branca","Calça Jeans","Jaqueta"]
precos= [49.90, 129.90,219.90]
#parte2
novo_produto = input("Digite o nome do novo produto")
novo_preco = float(input("Digite o preco do novo produto"))

produtos.append(novo_produto)
precos.append(novo_preco)

print("Estoque atualizado")
#parte3
for i,j in zip(produtos,precos):
 print(f"{produtos[i]} , {precos[i]}")

if len(produtos) > 0:
  produtos_removido = produtos.pop()
  preco_removido = precos.pop()
  print("Produto Removido:",produtos_removido)
  print("Preço Removido:", preco_removido)



else:
  print("Estoque vazio")

print("Produtos", produtos)
print("Precos", precos)





```
